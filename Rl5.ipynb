```python
import gymnasium as gym
import random


def run_random_agent(env_name):
    print("\n" + "=" * 50)
    print("Environment:", env_name)
    print("=" * 50)

    # Part A - Create Environment
    env = gym.make(env_name)

    # Part B - Reset environment
    state, info = env.reset()

    print("Initial State:", state)
    print("Action Space:", env.action_space)

    total_reward = 0
    step_count = 0

    # Part C - Continue until episode terminates
    while True:

        # Part D - Select random action
        action = env.action_space.sample()

        # Part E - Execute action
        next_state, reward, terminated, truncated, info = env.step(action)

        step_count += 1
        total_reward += reward

        print("\nStep:", step_count)
        print("State:", state)
        print("Action:", action)
        print("Reward:", reward)
        print("Next State:", next_state)

        # Move to next state
        state = next_state

        # Check whether episode has ended
        if terminated or truncated:
            break

    # Average reward per step
    average_reward = total_reward / step_count

    print("\n----- Results -----")
    print("Total Steps:", step_count)
    print("Total Reward:", total_reward)
    print("Average Reward:", round(average_reward, 4))

    env.close()


# Test 1 - CartPole
run_random_agent("CartPole-v1")

# Test 2 - MountainCar
run_random_agent("MountainCar-v0")

# Test 3 - FrozenLake
run_random_agent("FrozenLake-v1")
```

```python
import gymnasium as gym


def run_random_agent(env_name):

    env = gym.make(env_name)

    state, info = env.reset()

    total_reward = 0
    steps = 0

    while True:

        # Random action
        action = env.action_space.sample()

        # Environment interaction
        next_state, reward, terminated, truncated, info = env.step(action)

        total_reward += reward
        steps += 1

        state = next_state

        if terminated or truncated:
            break

    average_reward = total_reward / steps

    env.close()

    return steps, total_reward, average_reward


# Environments
environments = [
    "CartPole-v1",
    "MountainCar-v0",
    "FrozenLake-v1"
]

print("\n========== RANDOM AGENT RESULTS ==========")

print("\nEnvironment\t\tSteps\tTotal Reward\tAverage Reward")
print("-" * 65)

for env_name in environments:

    steps, total_reward, average_reward = run_random_agent(env_name)

    print(
        f"{env_name:<20}"
        f"{steps:<10}"
        f"{total_reward:<15.2f}"
        f"{average_reward:.4f}"
    )
```
