```python
# Assignment 1: Undiscounted and Discounted Returns

n = int(input("Enter number of time steps: "))

rewards = []

for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)

gamma = float(input("Enter discount factor (gamma): "))

# Undiscounted Return
undiscounted_return = sum(rewards)

# Discounted Return
discounted_return = 0

for i in range(n):
    discounted_return += (gamma ** i) * rewards[i]

# Display results
print("\nEntered Reward Values:", rewards)
print("Undiscounted Return:", undiscounted_return)
print("Discounted Return:", discounted_return)
```

```python
# Assignment 2: Deterministic Grid World MDP

# State Space
states = ["S0", "S1", "S2", "S3", "S4", "G"]

# Action Space
actions = ["up", "down", "left", "right"]

# Discount Factor
gamma = 0.9

# Goal and Obstacle
goal = "G"
obstacle = "S4"


# Reward Function
def reward(state, action, next_state):
    if next_state == goal:
        return 10
    elif next_state == obstacle:
        return -5
    else:
        return -1


# Transition Function
def transition(state, action):

    transitions = {
        "S0": {
            "up": "S0",
            "down": "S3",
            "left": "S0",
            "right": "S1"
        },

        "S1": {
            "up": "S1",
            "down": "S1",
            "left": "S0",
            "right": "S2"
        },

        "S2": {
            "up": "S2",
            "down": "G",
            "left": "S1",
            "right": "S2"
        },

        "S3": {
            "up": "S0",
            "down": "S3",
            "left": "S3",
            "right": "S3"
        },

        "S4": {
            "up": "S4",
            "down": "S4",
            "left": "S4",
            "right": "S4"
        },

        "G": {
            "up": "G",
            "down": "G",
            "left": "G",
            "right": "G"
        }
    }

    return transitions[state][action]
```

```python
# Import the Grid World environment

# The Grid World environment is defined in a previous cell and its components are directly available.
# import grid_env_det as env # This line is removed as the module does not exist

print("State Space:", states)
print("Action Space:", actions)
print("Discount Factor:", gamma)

# Get input from user
state = input("\nEnter current state: ")
action = input("Enter action: ")

# Find next state
next_state = transition(state, action)

# Find reward
r = reward(state, action, next_state)

# Display result
print("\nCurrent State:", state)
print("Action:", action)
print("Next State:", next_state)
print("Reward:", r)
```
