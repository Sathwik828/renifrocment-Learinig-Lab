```python

# ============================================================
# ASSIGNMENT 1: UNDISCOUNTED AND DISCOUNTED RETURNS
# ASSIGNMENT 2: GRID WORLD ENVIRONMENT AS AN MDP
# ============================================================

# ==================== ASSIGNMENT 1 ==========================

print("=" * 55)
print("ASSIGNMENT 1: RETURN CALCULATION")
print("=" * 55)

n = int(input("Enter the number of time steps: "))

rewards = []
for i in range(n):
    r = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(r)

gamma = float(input("Enter discount factor gamma (0 to 1): "))

undiscounted_return = sum(rewards)

discounted_return = sum(
    (gamma ** t) * rewards[t] for t in range(n)
)

print("\nEntered reward values:", rewards)
print("Undiscounted return:", undiscounted_return)
print("Discounted return:", round(discounted_return, 4))


# ==================== ASSIGNMENT 2 ==========================

print("\n" + "=" * 55)
print("ASSIGNMENT 2: GRID WORLD MDP")
print("=" * 55)


class GridWorld:
    def __init__(self):
        # a. State Space
        self.states = ["S0", "S1", "S2", "S3", "S4", "S5"]

        # b. Action Space
        self.actions = ["up", "down", "left", "right"]

        self.rows = 2
        self.cols = 3

        self.start_state = "S0"
        self.goal_state = "S2"
        self.obstacle = "S4"

        # c. Reward Function
        self.rewards = {
            "normal": -1,
            "goal": 10,
            "obstacle": -5
        }

        # d. Discount Factor
        self.gamma = 0.9

        # e. Transition Function
        self.transitions = {}

        for state in self.states:
            self.transitions[state] = {}
            for action in self.actions:
                next_state = self.get_next_state(state, action)
                self.transitions[state][action] = next_state

    def get_next_state(self, state, action):
        # Goal and obstacle are terminal/unavailable states
        if state in [self.goal_state, self.obstacle]:
            return state

        index = int(state[1])
        row = index // self.cols
        col = index % self.cols

        new_row, new_col = row, col

        if action == "up":
            new_row -= 1
        elif action == "down":
            new_row += 1
        elif action == "left":
            new_col -= 1
        elif action == "right":
            new_col += 1
        else:
            raise ValueError("Invalid action!")

        # Prevent movement outside the grid
        if not (0 <= new_row < self.rows and
                0 <= new_col < self.cols):
            return state

        next_state = f"S{new_row * self.cols + new_col}"

        # Prevent entry into obstacle
        if next_state == self.obstacle:
            return state

        return next_state

    # f. Estimate next state and reward
    def step(self, state, action):
        state = state.upper()
        action = action.lower()

        if state not in self.states:
            raise ValueError("Invalid state! Use S0 to S5.")

        if action not in self.actions:
            raise ValueError("Invalid action! Use up/down/left/right.")

        next_state = self.transitions[state][action]

        if next_state == self.goal_state:
            reward = self.rewards["goal"]

        elif next_state == state:
            index = int(state[1])
            row = index // self.cols
            col = index % self.cols

            new_row, new_col = row, col

            if action == "up":
                new_row -= 1
            elif action == "down":
                new_row += 1
            elif action == "left":
                new_col -= 1
            elif action == "right":
                new_col += 1

            if (0 <= new_row < self.rows and
                0 <= new_col < self.cols and
                f"S{new_row * self.cols + new_col}" == self.obstacle):
                reward = self.rewards["obstacle"]
            else:
                reward = self.rewards["normal"]

        else:
            reward = self.rewards["normal"]

        return next_state, reward


# Create the environment
env = GridWorld()

# Display MDP components
print("\nState Space:", env.states)
print("Action Space:", env.actions)
print("Start State:", env.start_state)
print("Goal State:", env.goal_state)
print("Obstacle State:", env.obstacle)
print("Reward Function:", env.rewards)
print("Discount Factor:", env.gamma)

# Display transition function
print("\nTransition Function:")
for state in env.states:
    print(state, ":", env.transitions[state])

# Get user input to estimate the next state
print("\n--- Next State Estimation ---")

current_state = input("Enter current state (S0 to S5): ").strip().upper()
action = input("Enter action (up/down/left/right): ").strip().lower()

try:
    next_state, reward = env.step(current_state, action)

    print("\nCurrent State:", current_state)
    print("Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward)
    print("Discount Factor:", env.gamma)

except ValueError as e:
    print("Error:", e)
```
