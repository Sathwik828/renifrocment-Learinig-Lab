```python

# Practical 9 - Assignment 1
# State and Action Values using Bellman Expectation Equations

gamma = 0.9

# Bellman equations:
# V(S0) = 0.9 * (-1 + gamma * V(S1)) + 0.1 * (-1)
# V(S1) = 0.9 * 100 + 0.1 * (-1 + gamma * V(S0))

# Simplified equations:
# V0 = -1 + 0.81 * V1
# V1 = 89.9 + 0.09 * V0

# Solve the simultaneous equations algebraically
V0 = (-1 + 0.81 * 89.9) / (1 - 0.81 * 0.09)
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q_S0_Down = -1 + gamma * V1
Q_S0_Right = -1

Q_S1_Right = 0.9 * (100 + gamma * 0) + \
             0.1 * (-1 + gamma * V0)

Q_S1_Up = -1 + gamma * V0

print("Assignment 1: Algebraic Solution")
print("--------------------------------")
print(f"V(S0)       = {V0:.2f}")
print(f"V(S1)       = {V1:.2f}")
print(f"Q(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q(S0, Right)= {Q_S0_Right:.2f}")
print(f"Q(S1, Right)= {Q_S1_Right:.2f}")
print(f"Q(S1, Up)   = {Q_S1_Up:.2f}")

# Verify Bellman expectation equations
V0_check = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
V1_check = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

print("\nVerification")
print(f"V(S0) check = {V0_check:.2f}")
print(f"V(S1) check = {V1_check:.2f}")
```

```python

# Practical 9 - Assignment 2
# Iterative Policy Evaluation in a Stochastic Environment

gamma = 0.9
theta = 0.000001

# Initialize state values
V0 = 0.0
V1 = 0.0
iterations = 0

# Iterative policy evaluation
while True:
    # Use previous values for both updates
    old_V0 = V0
    old_V1 = V1

    # S0 -> Wall (0.1), S1 (0.9)
    new_V0 = (
        0.1 * (-1 + gamma * 0)
        + 0.9 * (-1 + gamma * old_V1)
    )

    # S1 -> Goal (0.9), S0 (0.1)
    new_V1 = (
        0.9 * (100 + gamma * 0)
        + 0.1 * (-1 + gamma * old_V0)
    )

    # Calculate maximum change
    delta = max(
        abs(new_V0 - old_V0),
        abs(new_V1 - old_V1)
    )

    V0 = new_V0
    V1 = new_V1
    iterations += 1

    if delta < theta:
        break

# Calculate action values after convergence

# S0, Down: Wall (0.1), S1 (0.9)
Q_S0_Down = (
    0.1 * (-1 + gamma * 0)
    + 0.9 * (-1 + gamma * V1)
)

# S0, Right: Wall (1.0)
Q_S0_Right = -1 + gamma * 0

# S1, Right: Goal (0.9), S0 (0.1)
Q_S1_Right = (
    0.9 * (100 + gamma * 0)
    + 0.1 * (-1 + gamma * V0)
)

# S1, Up: S0 (1.0)
Q_S1_Up = -1 + gamma * V0

# Display results
print("Assignment 2: Iterative Policy Evaluation")
print("-----------------------------------------")
print("Number of iterations:", iterations)
print(f"V(S0)        = {V0:.2f}")
print(f"V(S1)        = {V1:.2f}")
print(f"Q(S0, Down)  = {Q_S0_Down:.2f}")
print(f"Q(S0, Right) = {Q_S0_Right:.2f}")
print(f"Q(S1, Right) = {Q_S1_Right:.2f}")
print(f"Q(S1, Up)    = {Q_S1_Up:.2f}")

# Verify policy consistency
print("\nPolicy Consistency Checks")
print("V(S0) == Q(S0, Down):",
      abs(V0 - Q_S0_Down) < theta)
print("V(S1) == Q(S1, Right):",
      abs(V1 - Q_S1_Right) < theta)
```
