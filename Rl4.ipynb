```python
import math

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of machines
M = 3

# Number of plays
N_play = 10

# Part B – Initialize N, R, Q and UCB
N = [0, 0, 0]          # Number of times each machine is played
R = [0, 0, 0]          # Total reward obtained from each machine
Q = [0.0, 0.0, 0.0]    # Estimated Q-value
UCB = [float('inf')] * M

# Total reward received by player
total_reward = 0

print("===== Upper Confidence Bound (UCB) Multi-Armed Bandit =====")
print("Machines:", machines)
print("Number of plays:", N_play)

# Play the machines 10 times
for play in range(1, N_play + 1):

    # Part C – Calculate UCB
    for i in range(M):
        if N[i] == 0:
            # Give unplayed machines infinite UCB
            # so that every machine gets selected at least once
            UCB[i] = float('inf')
        else:
            UCB[i] = Q[i] + math.sqrt((2 * math.log(play)) / N[i])

    # Select machine with highest UCB
    selected = UCB.index(max(UCB))
    machine = machines[selected]

    print("\nPlay", play)
    print("UCB values:")

    for i in range(M):
        if UCB[i] == float('inf'):
            print(machines[i], "= Infinity")
        else:
            print(machines[i], "=", round(UCB[i], 4))

    print("Selected Machine:", machine)

    # Part D – Reward Collection
    reward = float(input("Enter reward obtained: "))

    # Update number of plays
    N[selected] += 1

    # Update total reward
    R[selected] += reward
    total_reward += reward

    # Part E – Incremental Sample-Average Q-value
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    print("Updated Q-value of", machine, "=", round(Q[selected], 4))


# Part F – Display Final Results
print("\n\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tUCB\t\tQ-value")

for i in range(M):
    if N[i] == 0:
        UCB[i] = float('inf')
    else:
        UCB[i] = Q[i] + math.sqrt((2 * math.log(N_play)) / N[i])

    ucb_display = "Infinity" if UCB[i] == float('inf') else round(UCB[i], 4)

    print(
        machines[i],
        "\t", N[i],
        "\t", R[i],
        "\t", ucb_display,
        "\t\t", round(Q[i], 4)
    )

print("\nTotal Reward Received =", total_reward)
```
